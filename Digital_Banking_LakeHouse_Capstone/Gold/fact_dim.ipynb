```python
catalog = dbutils.widgets.get("catalog")
```

```python
# Read tables from the silver schema
# Dimension DataFrames
dim_accounts    = spark.table(f"{catalog}.silver.silver_accounts")
dim_branches    = spark.table(f"{catalog}.silver.silver_branches")
dim_customers   = spark.table(f"{catalog}.silver.silver_customers")

# Fact DataFrame
fact_transactions = spark.table(f"{catalog}.silver.silver_transactions")

# Write tables to the gold schema
dim_accounts.write.mode("overwrite").saveAsTable(f"{catalog}.gold.dim_accounts")
dim_branches.write.mode("overwrite").saveAsTable(f"{catalog}.gold.dim_branches")
dim_customers.write.mode("overwrite").saveAsTable(f"{catalog}.gold.dim_customers")
fact_transactions.write.mode("overwrite").saveAsTable(f"{catalog}.gold.fact_transactions")




```
