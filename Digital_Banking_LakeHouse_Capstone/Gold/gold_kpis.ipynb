```python
# KPI: Total Transaction Value — sum of all successful transactions

spark.sql(f"""
CREATE OR REPLACE VIEW dev.gold.kpi_total_transaction_value AS
SELECT
    ROUND(SUM(amount), 2) AS total_transaction_value
FROM dev.gold.fact_transactions
WHERE LOWER(transaction_status) = 'completed'
""")

# Preview the KPI
# spark.sql(f"SELECT * FROM dev.gold.kpi_total_transaction_value").display()
```

```python
# KPI: Total Transaction Count — Count of all successful transactions
spark.sql(f"""
CREATE OR REPLACE VIEW dev.gold.kpi_total_transaction_count AS
SELECT
    COUNT(*) AS total_completed_transaction_count
FROM dev.gold.fact_transactions
WHERE LOWER(transaction_status) = 'completed'
""")

# Preview the KPI
# spark.sql(f"SELECT * FROM dev.gold.kpi_total_transaction_count").display()
```

```python
# KPI: Total Value of Deposits and Withdrawals — breakdown by transaction_type
spark.sql(f"""
CREATE OR REPLACE VIEW dev.gold.kpi_total_value_by_withdrawal_deposit AS
SELECT
    transaction_type,
    ROUND(SUM(amount), 2) AS total_value
FROM dev.gold.fact_transactions
WHERE LOWER(transaction_status) = 'completed'
  AND LOWER(transaction_type) IN ('withdrawal', 'deposit')
GROUP BY transaction_type
""")

# Preview the KPI
# spark.sql("SELECT * FROM dev.gold.kpi_total_value_by_type").display()
```

```python
# KPI: Total Transaction Value by Month — track the trend of completed transactions over time
spark.sql(f"""
CREATE OR REPLACE VIEW dev.gold.kpi_transaction_value_by_month AS
SELECT
    DATE_TRUNC('month', transaction_date) AS transaction_month,
    ROUND(SUM(amount), 2) AS total_transaction_value,
    COUNT(*) AS transaction_count
FROM dev.gold.fact_transactions
WHERE LOWER(transaction_status) = 'completed'
GROUP BY DATE_TRUNC('month', transaction_date)
ORDER BY transaction_month
""")

# Preview the KPI
# spark.sql("SELECT * FROM dev.gold.kpi_transaction_value_by_month").
# display()
```
